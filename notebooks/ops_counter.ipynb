# KAN Ops Counter

Set either `MODEL_SPEC` (module:attr) or `TORCH_HUB_REPO`/`TORCH_HUB_MODEL`, then run the cells.


```python
from converted_KAN.ops_counter import count_ops
import importlib
import torch

# Option A: local module:attr (callable returning nn.Module)
MODEL_SPEC = None  # e.g. 'my_models:resnet20'

# Option B: torch.hub repo/model (like test_conversion.py)
TORCH_HUB_REPO = 'chenyaofo/pytorch-cifar-models'
TORCH_HUB_MODEL = 'cifar10_resnet20'

INPUT_SHAPE = (1, 3, 32, 32)
CONVERT_KAN = True
```

```python
if MODEL_SPEC:
    module_name, _, attr = MODEL_SPEC.partition(':')
    if not module_name or not attr:
        raise ValueError("MODEL_SPEC must be like 'module:attr'")

    mod = importlib.import_module(module_name)
    obj = getattr(mod, attr)
    model = obj() if callable(obj) else obj
else:
    if not TORCH_HUB_REPO or not TORCH_HUB_MODEL:
        raise ValueError('Set MODEL_SPEC or TORCH_HUB_REPO/TORCH_HUB_MODEL')
    model = torch.hub.load(TORCH_HUB_REPO, TORCH_HUB_MODEL, pretrained=True)

if CONVERT_KAN:
    from converted_KAN.convert import convert_to_kan
    model = convert_to_kan(model, inplace=False)
```

```python
counts = count_ops(model, INPUT_SHAPE, per_layer=True)
total = counts["total"]
per_layer = counts["per_layer"]

# Sort by total_arith desc
sorted_layers = sorted(per_layer.items(), key=lambda kv: kv[1]["total_arith"], reverse=True)
for name, metrics in sorted_layers:
    print(f"{name}: {metrics}")
```
